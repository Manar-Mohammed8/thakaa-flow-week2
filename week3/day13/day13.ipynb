**مثال من حياتي:** متوسط استخدامي للجوال 4 ساعات باليوم لو شفت هالرقم بس بتوقعه ثابت، لكن الواقع فيه أيام أستخدمه ساعة وحدة وأيام ثانية 8 ساعات الرسم الخطي يوضح لي هالتذبذب بين الأيام عكس المتوسط اللي يخفيه، وهذا بالضبط اللي يصير في المجموعات الأربعة؛ أرقامها متشابهة لكن رسماتها تكشف أن توزيعها مختلف تماما

### السؤال 2: اختر الرسمة

| السؤال | نوع الرسم | لماذا؟ |

### السؤال 3: أجزاء الرسمة

- **Figure:** هي الصفحة أو اللوحه الكاملة اللي تحتوي على الرسم كله
- **Axes:** هي منطقة الرسم نفسها داخل الـ Figure الي تنعرض فيها البيانات والمحاور

<p dir="rtl"><b>وظيفة كل عنصر:</b></p>

- **العنوان (Title):** يوضح موضوع الرسم بشكل عام
- **عنوان المحور الأفقي (X-axis Label):** يوضح ماذا تمثل قيم المحور الأفقي
- **عنوان المحور العمودي (Y-axis Label):** يوضح ماذا تمثل قيم المحور العمودي ووحدة القياس
- **Legend:** يوضح معنى كل لون أو خط أو رمز إذا كان فيه أكثر من مجموعة بيانات

**ايش يصير لو أرسلت رسمة لمديري بدونها؟**
ماراح يفهم الرسمة عن ايش تتكلم لأن العنوان مو موجود، ولا بيعرف وش تعني الأرقام أو وحداتها بسبب غياب عناوين المحاور، واذا كان فيه أكثر من خط أو لون ما بيعرف كل واحد وش يمثل بدون الـ Legend، وهذا كله بيخليه يفسر البيانات بشكل غلط

</div>

### توقعي (Bar) : قبل التشغيل
أتوقع مشروب اللاتيه يحقق أعلى إجمالي مبيعات، لان مبيعاته اليومية أعلى من بقية المشروبات في معظم الأيام

</div>

```python
import matplotlib.pyplot as plt

# البيانات
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]

drinks = ["Latte", "Espresso", "Mocha", "Tea"]

# حساب المجاميع
latte_total = sum(latte)
espresso_total = sum(espresso)

# قائمة الإجماليات
totals = [latte_total, espresso_total, 145, 96]

# رسم الأعمدة
plt.bar(drinks, totals)

# كتابة الرقم فوق كل عمود
for i, value in enumerate(totals):
    plt.text(i, value + 5, value, ha="center")

# العنوان والمحاور
plt.title("Total Drink Sales")
plt.xlabel("Drinks")
plt.ylabel("Sales")

plt.show()
```

### توقعي (Histogram)
أتوقع أن التوزيع يكون متقارب وممتد بشكل أفقي، وأن أغلب القيم تكون في المنتصف، وأتوقع أن المتوسط والوسيط يكونان متقاربين لأن أغلب القيم متجمعة في نفس المنطقة

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")
df = df.drop_duplicates()
df.head()
```

```python
plt.hist(df["charges"], bins=30)

plt.axvline(df["charges"].mean(), color="red", label="Mean")
plt.axvline(df["charges"].median(), color="green", label="Median")

plt.title("Distribution of Insurance Charges")
plt.xlabel("Charges")
plt.ylabel("Count")
plt.legend()

plt.show()

print("Mean:", df["charges"].mean())
print("Median:", df["charges"].median())
```

<div dir="rtl">

### ملاحظتي
كان توقعي غير صحيح، لأن المتوسط أكبر من الوسيط (المتوسط 13279.12 والوسيط 9386.16). وهذا يدل على أن التوزيع منحرف لليمين، بسبب وجود قيم مرتفعة رفعت المتوسط أكثر من الوسيط

</div>

### توقعي (Scatter)
أتوقع أن تكون هناك علاقة طردية بين العمر والتكاليف، يعني كلما زاد العمر زادت تكاليف التأمين. وأتوقع أن تكون تكاليف المدخنين أعلى من غير المدخنين، لذلك تظهر نقاط المدخنين في مستوى أعلى من نقاط غير المدخنين

```python
non_smoker = df[df["smoker"] == "no"]
smoker = df[df["smoker"] == "yes"]

plt.scatter(non_smoker["age"], non_smoker["charges"], label="Non-smoker")
plt.scatter(smoker["age"], smoker["charges"], label="Smoker")

plt.title("Age vs Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Charges")
plt.legend()

plt.show()
```

<div dir="rtl">

### ملاحظتي (Scatter)

</div>

<div dir="rtl">

### توقعي (Bar)
أتوقع أن متوسط تكاليف المدخنين أعلى من غير المدخنين لأن التدخين قد يسبب مشاكل صحية تزيد من تكاليف التأمين. وأتوقع أن المناطق الأربع تكون متباعده في متوسط التكاليف

</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

smoker_mean = df.groupby("smoker")["charges"].mean()
region_mean = df.groupby("region")["charges"].mean()

axes[0].bar(smoker_mean.index, smoker_mean.values)
axes[0].set_title("Mean Charges by Smoker")
axes[0].set_xlabel("Smoker")
axes[0].set_ylabel("Mean Charges")

axes[1].bar(region_mean.index, region_mean.values)
axes[1].set_title("Mean Charges by Region")
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Mean Charges")

plt.tight_layout()
plt.show()

print(smoker_mean)
print(region_mean)
```

طلع توقعي صح بالنقطة الأولى وغلط بالثانية؛ متوسط تكاليف المدخنين (32050) طلع أعلى بكثير من غير المدخنين (8441) مثل ما توقعت لكن بالنسبة للمناطق كنت متوقعة إنها متباعدة، وطلعت العكس تماما ومتشابهة (بين 12347 و14735)، بحيث كانت منطقة southeast هي الأعلى ومنطقة southwest هي الأقل

### محطة أرصاد المريخ
محطة أرصاد على المريخ أرسلت قراءات 60 يومًا مريخيًا (Sol)، ومهمتي اختيار أفضل نافذة لاطلاق مركبة

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)
temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)
storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1
temp[storm == 1] -= 12
wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)
```

<div dir="rtl">

### توقعي
أتوقع أيام العواصف تكون أبرد من باقي الأيام، لأن الكود ينقص 12 درجة من الحرارة في أيام العواصف

</div>

```python
plt.plot(sol, temp, label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Dust storm")
plt.title("Temperature per Sol")
plt.xlabel("Sol")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

<div dir="rtl">

### ملاحظتي
كان توقعي صح جزئيا لأن الكود ينقص 12 درجة في أيام العواصف لكن لاحظت أن النقاط الحمراء ما كانت كلها في أبرد مكان، وظهرت في أماكن مختلفة على الخط بسبب تغير الحرارة مع الوقت وبعض نقاط العواصف كانت عند أقل درجة تقريبًا (-87)، بينما نقاط أخرى كانت أعلى مثل -55 تقريبًا

</div>

<div dir="rtl">

### توقعي (annotate)
أتوقع أن أبرد يوم يكون تقريبًا عند Sol 46 أو 47، وأن درجة حرارته تكون حوالي -87°C لأنه يظهر كأقل نقطة في الرسم

</div>

```python
mean_temp = temp.mean()
coldest_idx = temp.argmin()
coldest_sol = sol[coldest_idx]
coldest_temp = temp[coldest_idx]

plt.plot(sol, temp, label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", label="Dust storm")
plt.axhline(mean_temp, color="green", linestyle="--", label="Mean")
plt.annotate(f"Coldest: Sol {coldest_sol}, {coldest_temp}°C",
             xy=(coldest_sol, coldest_temp),
             xytext=(coldest_sol - 25, coldest_temp + 8),
             arrowprops=dict(arrowstyle="->"))
plt.title("Temperature per Sol")
plt.xlabel("Sol")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()

print("Mean:", mean_temp)
print("Coldest:", coldest_sol, coldest_temp)
```

<div dir="rtl">

### ملاحظتي (annotate)
كان توقعي قريب من الصح؛ درجة الحرارة كانت حوالي -87.5°C، لكن أبرد يوم طلع Sol 48 مو 46 أو 47. ولاحظت أن متوسط الحرارة كان -60.67°C، وأبرد يوم كان أقل منه بـ 27 درجة تقريبًا، وطلع بالفعل من أيام العواصف

</div>

<div dir="rtl">

### توقعي (subplots)
أتوقع أنه في أيام العواصف تكون سرعة الرياح أعلى من باقي الأيام، وتظهر كارتفاعات واضحة في رسم الرياح

</div>

### توقعي (أفضل نافذة إطلاق)
أتوقع أن أفضل أيام للاطلاق تكون الأيام اللي تكون حرارتها بين -45 و -50 تقريبًا، لأن الجو فيها أدفأ من باقي الأيام وما تظهر فيها عواصف، لذلك أتوقع أن تكون الرياح أقل وتكون مناسبة للإطلاق

```python
best_mean = None
best_start = None

for i in range(len(sol) - 4):
    if storm[i:i+5].sum() == 0:
        m = temp[i:i+5].mean()
        if best_mean is None or m > best_mean:
            best_mean = m
            best_start = i

print("Best window: Sol", sol[best_start], "to", sol[best_start + 4])
print("Mean temp:", best_mean)
```
