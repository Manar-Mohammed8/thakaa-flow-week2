### السؤال 2: كيف يعمل Streamlit؟

في Streamlit كل ما المستخدم يغير شي بالواجهة مثل يحرك الـ slider ينعاد تشغيل الملف كامل من البداية. ولو البرنامج يقرأ ملف CSV كبير، راح يضطر يقرأه من جديد في كل مرة وهذا يثقل التطبيق ويخليه بطيء. فايدة `@st.cache_data` إنها تحفظ البيانات بالذاكرة المؤقتة من أول مرة يشتغل فيها، عشان ما يرجع يقرأ الملف كل شوي ويصير التطبيق أسرع بكثير

### السؤال 3: لماذا تطبيق وليس Notebook فقط؟

محلل البيانات يحتاج تطبيق ومو بس Notebook لأن التطبيق أسهل بكثير للناس اللي ما تعرف بايثون. المدير أو أي مستخدم يقدر يتفاعل مع البيانات ويشوف النتائج من واجهة مرتبة وبسيطة بدون ما يضطر يشغل أكواد أو يفهم في البرمجة. مثلاً في تطبيق القروض المدير يقدر يفتح الرابط ويغير المنطقة (Urban أو Rural أو Semiurban) من القائمة ويشوف نسبة الموافقة مباشرة بنفسه، بدون ما يثبت بايثون أو يطلب مني أعدل الكود كل مرة

## Error Log (Task 3)

ما ظهر عندي أي خطأ وقت النشر على Streamlit Cloud، وفتح الرابط واشتغلت الفلاتر معي من الجوال تمام

لو نسيت رفع ملف `requirements.txt`: كان التطبيق بيشل في السحابة وبيطلع لي رسالة خطأ زي `ModuleNotFoundError` (مثلاً لمكتبة seaborn)، لأن السحابة ما تكون عارفة وش المكتبات اللي يحتاجها التطبيق، وهي ما تثبتها إلا من هذا الملف

أخطاء ظهرت معي قبل النشر (محلياً):
- خطأ ``  `FileNotFoundError` لملف `loan_clean.csv`، والسبب إن الملف ما كان بنفس مجلد `app.py`. وحليتها إني نسخت الملف للمجلد، بعدين رفعته على قيت هوب (GitHub) مع `app.py` لان السحابة ما تشوف إلا الملفات اللي في الريبو
- ملف `requirements.txt` انحفظ في المجلد الغلط، ونقلته للمجلد الصحيح `loan-dashboard`
