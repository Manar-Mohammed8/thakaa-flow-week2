**السؤال 1: رأيي قبل الحساب**

أرى أن هذا المتوسط قد يكون مضللًا، لأن معظم الرواتب بين 6000 و8000 ريال، ويوجد راتب مرتفع جدًا يؤثر على المتوسط

</div>

```python
import numpy as np
salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])
```

<div dir="rtl">

**توقعي (قبل التشغيل):** اتوقع الوسيط سيكون حوالي 7200 ريال لانه في منتصف الرواتب بعد ترتيبها. واتوقع راتب 95000 ريال سيكون هو الـ Outlier لانه أعلى بكثير من بقية الرواتب

</div>

```python
mean_salary = salaries.mean()
median_salary = np.median(salaries)
Q1 = np.percentile(salaries, 25)
Q3 = np.percentile(salaries, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Mean:", mean_salary)
print("Median:", median_salary)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = salaries[(salaries < lower_bound) | (salaries > upper_bound)]
print("Outliers:", outliers)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا. الوسيط 7200 والمتوسط 19600، والـ Outlier الوحيد هو راتب 95000 لانه اعلى من الحد الأعلى 9250

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** اتوقع المتوسط سيتغير اكثر من الوسيط، لان المتوسط يتأثر كثيرًا بالقيم الكبيرة جدًا مثل راتب المدير (95000 ريال)، بينما الوسيط يعتمد على القيمة الموجودة في المنتصف لذلك يكون أقل تأثرًا

</div>

```python
no_manager = salaries[salaries < 95000]

mean_no_manager = no_manager.mean()
median_no_manager = np.median(no_manager)

print("Mean without manager:", mean_no_manager)
print("Median without manager:", median_no_manager)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا. المتوسط تغير من 19600 الى 7033، بينما الوسيط تغير من 7200 إلى 7100 فقط. لذلك المتوسط حساس للقيم الشاذة، أما الوسيط فمقاوم لها

**توقعي (قبل التشغيل):** اتوقع أن charges فيها Outliers أكثر من bmi، لان الفرق بين المتوسط والوسيط فيها كبير، وهذا يدل على وجود قيم مرتفعة جدًا تؤثر على المتوسط أكثر من بقية البيانات
</div>

```python
def iqr_outliers(col):
    Q1 = np.percentile(col, 25)
    Q3 = np.percentile(col, 75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = col[(col < lower) | (col > upper)]

    return Q1, Q3, IQR, lower, upper, outliers


for name, col in [("charges", charges), ("bmi", bmi)]:
    Q1, Q3, IQR, lower, upper, outliers = iqr_outliers(col)

    percentage = (len(outliers) / len(col)) * 100

    print(f"\n{name}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower bound: {lower:.2f}")
    print(f"Upper bound: {upper:.2f}")
    print(f"Outliers: {len(outliers)}")
    print(f"Percentage: {percentage:.2f}%")
```

<div dir="rtl">

**بعد التشغيل:** توقعي طلع صحيحًا، لأن charges فيها 139 Outlier و bmi فيها 9 فقط. أعتقد أن bmi فيها Outliers قليلة لان معظم قيمها متقاربة ولا توجد فيها قيم مرتفعة جدًا مثل التكاليف

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن أرى في أعلى 5 قيم من `charges` أرقامًا مرتفعة جدًا ومبتعدة عن أغلب القيم، بينما اعلى 5 قيم في `bmi` ستكون مرتفعة لكن متقاربة أكثر من بعضها

</div>

```python
for name, col in [("charges", charges), ("bmi", bmi)]:
    sorted_col = np.sort(col)
    top5 = sorted_col[-5:]
    print(name, top5)
```

**بعد التشغيل:** توقعي كان صحيحًا، لأن أعلى قيم `charges` كانت مرتفعة جدًا (من 55 ألف إلى 63 ألف تقريبًا) وبعيدة عن الوسيط، بينما أعلى قيم `bmi` كانت مرتفعة لكنها متقاربة من بعضها. هذا يدل على ان القيم الشاذة في `charges` أقوى وأكثر تأثيرًا من القيم الشاذة في `bmi`.

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن عدد الـ Outliers باستخدام z-score في `charges` سيكون اقل من 139، لان وجود قيم مرتفعة كثيرة يرفع المتوسط والانحراف المعياري، فيصبح من الأصعب أن تتجاوز القيم الشرط `|z| > 3`.

</div>

```python
for name, col in [("charges", charges), ("bmi", bmi)]:
    z = (col - col.mean()) / col.std()
    count = (np.abs(z) > 3).sum()
    print(name, count)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، `charges` طلعت 7 فقط بالـ z-score مقابل 139 بالـ IQR. السبب أن الـ IQR يعتمد على Q1 وQ3، والقيم الكبيرة جدًا لا تغيّرهما، لأنها فقط تأخذ مكانها في آخر الترتيب. أما الـ z-score فيعتمد على المتوسط والانحراف المعياري، والقيم الكبيرة ترفعهما، فيصير الحد `|z| > 3` ابعد، وتقل القيم التي تتجاوزه

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن الفرق بين المتوسطين سيكون كبيرًا، لأن القيم الشاذة المرتفعة في `charges` ترفع المتوسط بشكل واضح، مثلما رفع راتب المدير المتوسط في Task 1

</div>

```python
Q1, Q3, IQR, lower, upper, outliers = iqr_outliers(charges)

clean = charges[(charges >= lower) & (charges <= upper)]

print("Mean with outliers:", charges.mean())
print("Mean without outliers:", clean.mean())
print("Difference:", charges.mean() - clean.mean())
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لأن المتوسط انخفض من 13,270.42 إلى 9,927.75 بعد إزالة الـ Outliers، وهذا يدل على أن القيم الشاذة كانت ترفع المتوسط بشكل ملحوظ. كما أن المتوسط بدون الـ Outliers (9,927.75) أصبح أقرب إلى الوسيط (9,382)، مما يعني أنه أصبح يمثل البيانات بشكل أفضل.

<div dir="rtl">

**توقعي (قبل التشغيل):** اتوقع المدخنين قليل بين كل العملاء، واتوقع ان نسبة المدخنين بين الـ Outliers ستكون اعلى من النسبة العامة، لان التدخين قد يرفع تكاليف التأمين


</div>

```python
smoker = np.genfromtxt("insurance.csv", delimiter=",", skip_header=1, usecols=4, dtype=str)

is_smoker = smoker == "yes"
is_outlier = (charges < lower) | (charges > upper)

print("All customers:", is_smoker.mean() * 100)
print("Outliers:", is_smoker[is_outlier].mean() * 100)
print("Non-outliers:", is_smoker[~is_outlier].mean() * 100)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لان نسبة المدخنين بين الـ Outliers (97.84%) كانت اعلى بكثير من النسبة العامة (20.48%)، مما يدل على ارتفاع التكاليف له سبب حقيقي وليس مجرد خطأ في البيانات

**القرار:** سأختار التحقق والإبقاء على هذه القيم بدل حذفها، لأنها تبدو مرتبطة بصفة مهمة (التدخين) وقد تمثل عملاء حقيقيين ذوي تكاليف مرتفعة.

## Task 5

```python
import numpy as np
rng = np.random.default_rng(7)
brightness = 1000 + rng.normal(0, 1, 240)
for start in [30, 102, 174]:
    brightness[start:start + 5] -= 20
```

<div dir="rtl">

**توقعاتي (قبل التشغيل):**

1. اتوقع ان المتوسط والوسيط سيكونان متقاربين، لان عدد القراءات المنخفضة قليل (15 من أصل 240 قراءة) فلا اتوقع ان تؤثر كثيرًا على البيانات كلها

2. أتوقع أن عدد القراءات التي تحقق `z < -3` سيكون حوالي 15 قراءة، واتوقع ان تنكشف معظم الانخفاضات او كلها لأنها أكبر بكثير من التذبذب الطبيعي في البيانات

</div> 

```python
mean_b = brightness.mean()
median_b = np.median(brightness)

print("Mean:", mean_b)
print("Median:", median_b)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لأن المتوسط (998.60) والوسيط (999.84) كانا متقاربين. المتوسط أصغر قليلًا لأن الانخفاضات سحبته الى الأسفل.

</div>

```python
z = (brightness - brightness.mean()) / brightness.std()

mask = z < -3

print("Count:", mask.sum())

hours = np.arange(240)[mask]

print("Hours:", hours)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لأن الـ z-score اكتشف جميع الانخفاضات وعددها 15. ونجح هنا لأن الانخفاضات كانت كبيرة مقارنة بالتذبذب الطبيعي الصغير في البيانات

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن بدايات العبور ستكون 3 فقط، لأن الانخفاضات متجمعة في ثلاث فترات منفصلة.

**الشرح:** `np.diff` يحسب الفرق بين كل ساعة والتي بعدها، وإذا كان الفرق أكبر من 1 فهذه بداية عبور جديد، و`hours[0]` اول بداية

</div>

```python
diffs = np.diff(hours)
starts = np.concatenate([[hours[0]], hours[1:][diffs > 1]])
print("Starts:", starts)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لأن البدايات كانت 30 و102 و174 كما توقعت. وهي تطابق البدايات الموجودة في كود الورقة

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن الفواصل الزمنية بين العبورات ستكون متساوية

</div>

```python
periods = np.diff(starts)

print("Hours:", periods)
print("Days:", periods / 24)
```

<div dir="rtl">

**بعد التشغيل:** توقعي كان صحيحًا، لان الفاصل بين كل عبورين كان 72 ساعة (3 أيام). وهذا يعني ان الكوكب يكمل دورة حول نجمه كل 3 أيام

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):** أتوقع أن مقدار الانخفاض قريب من 20 وحدة، وأن نسبته صغيرة جدًا مقارنة بسطوع النجم

</div>

```python
normal_brightness = brightness[~mask].mean()
transit_brightness = brightness[mask].mean()

drop = normal_brightness - transit_brightness
drop_percent = (drop / normal_brightness) * 100

print("Drop:", drop)
print("Drop percent:", drop_percent)
```
