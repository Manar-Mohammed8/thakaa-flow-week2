**السؤال 1: الفرق بين Feature Engineering و Feature Selection**

* **Feature Engineering** = انشاء او توليد Features جديده من البيانات الموجوده
* **Feature Selection** = اختيار افضل الـ Features الموجوده وحذف غير المفيد منها

**مثال من كافيه**

* **Target:** التنبؤ هل العميل سيشتري قهوة مرة ثانيه خلال الاسبوع القادم او لا

**1) Feature Engineering**

لدينا الاعمده: عدد الزيارات، واجمالي المبلغ المصروف

استطيع انشاء Feature جديده اسمها: `Average_Spend_Per_Visit` = اجمالي المبلغ المصروف ÷ عدد الزيارات

هذه ما كانت موجوده في البيانات الاصلية، وقمنا بتوليدها من اعمده موجوده

**2) Feature Selection**

نفترض ان لدينا عمود اسمه `Customer_ID`، بحذفه لانه رقم تعريفي للعميل ولا يساعد بالغالب في التنبؤ اذا كان العميل بيعود للشراء او لا، وممكن يضيف ضوضاء للنموذج

**الخلاصة**

* Feature Engineering: انشاء عمود جديد مثل `Average_Spend_Per_Visit`
* Feature Selection: حذف او اختيار الاعمدة الموجوده مثل حذف `Customer_ID` لانه غير مفيد للتنبؤ

### السؤال 2: عائلات اختيار الـ Features

| العائلة | الفكرة بجملة من عندي | مثال لطريقة | ميزة | عيب |
|---|---|---|---|---|
| Filter | تختار الـ Features المهمه عن طريق مقاييس إحصائية قبل بناء الموديل | Correlation (الارتباط) | سريعة وسهلة وما تحتاج تدريب موديل | ممكن تتجاهل العلاقة بين اكثر من Feature. |
| Wrapper | تجرب مجموعات مختلفه من الـ Features وتشوف اي مجموعة تعطي افضل نتيجه مع الموديل | Recursive Feature Elimination (RFE) | تساعد في الوصول لمجموعة Features مناسبه وتعطي نتائج جيده | تأخذ وقت اطول وتحتاج حسابات اكثر |
| Embedded | الموديل يحدد الـ Features المهمة بنفسه أثناء التدريب. | Lasso Regression | توفر الوقت وتعطي أداء جيد. | تعتمد على نوع الموديل المستخدم |

### لماذا نستخدم Filter فقط هذا الأسبوع؟

لانها اسهل واسرع من الطرق الثانية، وما تحتاج نبني او ندرب موديل. وبهالطريقه نقدر نفهم فكرة اختيار الـ Features ونركز على البيانات اول، وبعدها نتعلم الطرق الثانية مثل Wrapper و Embedded

### السؤال 3: Data Leakage

| العمود | يجوز أو لا يجوز | السبب |
|---|---|---|
| `Disbursement_Date` | لا يجوز | لاننا نعرف تاريخ صرف القرض بعد الموافقة عليه، يعني المعلومة ما تكون متوفرة وقت اتخاذ القرار، وهذا يعتبر Data Leakage |
| `Customer_Since_Years` | يجوز | لان المعلومة تكون معروفة قبل الموافقة على القرض، وممكن تساعد الموديل في التنبؤ بالنتيجة |

### ماذا سيحدث لو استخدمنا الأعمدة المسببة للـ Data Leakage؟

ممكن تظهر دقة الموديل عالية جدا وقت التدريب والاختبار، لكنه في الحقيقة يعتمد على معلومات ما تكون متوفرة وقت التنبؤ. لذلك عند استخدامه في الواقع ممكن ينخفض أداؤه بشكل كبير وتكون نتائجه غير موثوقة

### الخلاصة

- `Customer_Since_Years`: يجوز
- `Manager_Final_Decision`: لا يجوز (Data Leakage)

### السؤال 4: هل «ارتباط قريب من الصفر يعني ان العمود ما يفيد»؟

لا، مو شرط يكون الكلام هذا صحيح دايم

لان الـ Correlation يقيس العلاقة الخطية فقط، وممكن يكون الارتباط قريب من الصفر مع ان الـ Feature تكون مفيدة ولها تاثير

### مثال من حياتي: ساعات النوم والنشاط

- اذا نمت ساعات قليلة جدا، يكون نشاطي منخفض
- اذا نمت عدد ساعات مناسب (مثل 7-8 ساعات)، يكون نشاطي مرتفع
- اذا نمت ساعات كثيرة جدا، ممكن احس بالخمول ويقل نشاطي مرة ثانية

يعني العلاقة ترتفع بالبداية وبعدها تنخفض، وما تكون على شكل خط مستقيم

في هالحالة ممكن يظهر الـ Correlation قريب من الصفر، لان الجزء الي ترتفع فيه العلاقة والجزء اللي تنخفض فيه يلغون بعض عند حساب الارتباط

لكن هذا ما يعني ان `Hours_Sleep` غير مفيد، بالعكس ممكن يكون له تاثير كبير على النشاط

لذلك مو كل Feature يكون ارتباطها قريب من الصفر يعني انها غير مفيدة، لان الـ Correlation ممكن ما يكتشف العلاقات غير الخطية ويعطينا فكرة غير صحيحة عن اهمية الـ Feature

وممكن بعد ارسم Scatter Plot بين الـ Feature والـ Target عشان اشوف شكل العلاقة بشكل اوضح، لان الرسم يساعدني اكتشف العلاقات اللي ممكن الارتباط ما يوضحها

#### توقعاتي قبل التشغيل

1. **السجل الائتماني (`Credit_History`):**  
   اتوقع ان يكون هو اللي فيه اكبر فجوة بين اعلى واقل نسبة موافقة لانه مرتبط مباشرة بسجل العميل الائتماني

2. **فئات الدخل (`Income_bin`):**  
   اتوقع ان تكون الفجوة فيه هي الاقل لاننا شفنا قبل كذا ان دخل المتقدم لحاله ما غير نسبة الموافقة بشكل واضح

</div>

```python
cat_cols = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Credit_History",
    "Property_Area",
    "Has_Coapplicant",
    "Payment_bin",
    "Income_bin"
]

results = []

for col in cat_cols:
    rates = df3.groupby(col, observed=False)["Loan_Status"].mean() * 100
    gap = rates.max() - rates.min()
    results.append({
        "column": col,
        "max_rate": rates.max(),
        "min_rate": rates.min(),
        "gap": gap
    })

gap_table = pd.DataFrame(results).sort_values("gap", ascending=False)

print(gap_table)
```

#### ملاحظاتي بعد التشغيل

1. **السجل الائتماني (`Credit_History`):**

#### توقعي قبل التشغيل

اتوقع ان `probe_id` يكون معرف لان كل صف يمثل مسبار مختلف لذلك اتوقع ان كل قيمه بتكون مختلفة

واتوقع بعد ان `mission_code` يحتوي على قيمتين مختلفتين مو قيمة واحدة بس لانه ممكن يمثل اكثر من مهمة او نوع مهمة

</div>

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 300
life = rng.integers(0, 2, n)
probes = pd.DataFrame({
    "probe_id": [f"P{i:03d}" for i in range(n)],
    "mission_code": "AX-7",
    "methane_ppm": np.round(5 + life * 12 + rng.normal(0, 3, n), 1),
    "temperature_c": np.round(-40 + life * 25 + rng.normal(0, 15, n), 1),
    "radiation": np.round(rng.normal(100, 10, n), 1),
    "humidity": np.round(20 + life * 6 + rng.normal(0, 8, n), 1),
    "life_detected": life,
})
probes["temperature_f"] = probes["temperature_c"] * 9 / 5 + 32
print(probes.shape)
probes.head()
```

```python
print(probes.nunique())
```

كان توقعي صح بالنسبة لـ `probe_id` حيث ظهر ان عنده **300 قيمة مختلفة** وهو نفس عدد الصفوف في البيانات لذلك يعتبر معرف فريد لكل سجل وما يضيف فائدة للتنبؤ فعشان كذا بنحذفه

لكن `mission_code` فتوقعت انه يحتوي على قيمتين مختلفتين بس النتيجة اظهرت ان عنده قيمة وحدة بس لذلك كان توقعي غلط وبما انه ثابت في كل الصفوف يفرق بين مسبار فيه حياة ومسبار ما فيه، لأن قيمته نفسها لكل المسابير بين السجلات وبنحذفه بعد

ولاحظت بعد ان `temperature_c` و `temperature_f` عندهم **246 قيمة مختلفة** لكل واحد منهم واعتقد ان السبب هو ان `temperature_f` محسوب مباشرة من `temperature_c` باستخدام معادلة تحويل الوحدات لذلك يحملون نفس المعلومة تقريبا بس بوحدتين مختلفتين وهذا ممكن يخلي واحد منهم مكرر للثاني

#### الحساسات المستبعدة

* **`radiation`:** استبعدته لان ارتباطه مع `life_detected` كان الاضعف بين الحساسات الباقية (**0.110** بس) وكمان متوسطه تغير بشكل محدود من **98.74** الى **100.96** وهذا يدل على انه ما يميز بوضوح بين وجود الحياة وعدم وجودها مقارنة بباقي الحساسات
* **`temperature_f`:** مستبعد لانه مكرر من `temperature_c` (ارتباط **1.0**) وما يضيف اي معلومة جديدة

#### الطريقة البديلة لو كان عندي وقت لتجربة موديل حقيقي
