**المشكلة:** التنبؤ بمستوى الزحام المروري في شوارع الرياض للتحكم الذكي في الإشارات المرورية.

**Features (5):**

1. وقت اليوم (الساعة).
2. يوم الأسبوع.
3. الحي أو موقع التقاطع.
4. حالة الطقس.
5. هل اليوم إجازة أم يوم عمل.

**عمود متاح لكن لن أدخله للموديل:** رقم التقاطع (Intersection ID).

**السبب:** لأن رقم التقاطع مجرد مُعرّف للموقع ولا يصف حالة المرور أو مستوى الزحام، بينما موقع التقاطع أو الحي يعطي معلومات مفيدة عن المنطقة وقد يساعد في التنبؤ بالازدحام.

```python
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
sales = np.array([1240, 980, 1105, 1320, 1875, 2410, 2150])
```

```python
print("Total:", sales.sum())
print("Mean:", sales.mean())
print("Median:", np.median(sales))
print("Max:", sales.max())
print("Min:", sales.min())
print("Std:", sales.std())
```

```python
best = sales.argmax()
worst = sales.argmin()

print("Best day:", days[best], sales[best])
print("Worst day:", days[worst], sales[worst])
```

```python
mask = sales > sales.mean()

print("Above average:", mask)
print("Days above average:", mask.sum())
print("Their sales:", sales[mask])
```

```python
weekend_sales = sales[-2:]

print("Weekend sales:", weekend_sales)

weekend_total = weekend_sales.sum()
week_share = (weekend_total / sales.sum()) * 100

print("Weekend total:", weekend_total)
print("Share:", round(week_share, 1), "%")
```

```python
after_tax = sales * 1.15
target = sales * 1.10

print("After tax:", after_tax)
print("Next week target:", target)
```

```python
changes = np.diff(sales)

print("Changes:", changes)

biggest_jump = changes.argmax()

print("Biggest jump:", changes[biggest_jump])
print("Day:", days[biggest_jump + 1])
```

## Task 4

```python
branches = ["Riyadh", "Jeddah", "Dammam"]
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
sales = np.array([
    [1240,  980, 1105, 1320, 1875, 2410, 2150],
    [ 910,  870,  990, 1010, 1430, 1980, 2240],
    [ 640,  720,  680,  750,  990, 1210, 1080],
])
```

<div dir="rtl">

**توقعي (قبل التشغيل):** شكل المصفوفة `(3, 7)`.

- الرقم الأول (3) يمثل عدد الصفوف، أي عدد الفروع
- الرقم الثاني (7) يمثل عدد الأعمدة، أي عدد أيام الأسبوع

</div>

```python
print(sales.shape)
```

" توقعي طلع صحيح "

```python
print("Jeddah Thursday:", sales[1, 4])

print("Dammam week:", sales[2, :])

print("Friday and Saturday all branches:")
print(sales[:, 5:])
```

<div dir="rtl">

**توقعي (قبل التشغيل):**

- `branch_totals` سيحتوي على 3 ارقام لأنه يجمع مبيعات كل فرع خلال الاسبوع
- `day_totals` سيحتوي على 7 أرقام لانه يجمع مبيعات جميع الفروع لكل يوم

</div>

```python
branch_totals = sales.sum(axis=1)
day_totals = sales.sum(axis=0)

print("Branch totals:", branch_totals)
print("Day totals:", day_totals)
```

<div dir="rtl">

**الفرق بين `axis=0` و `axis=1`:**

- `axis=1` يجمع القيم داخل كل صف، ليعطينا مجموع مبيعات كل فرع وظهرت 3 أرقام لان لدينا 3 فروع
- `axis=0` يجمع القيم داخل كل عمود، ليعطينا مجموع المبيعات لكل يوم وظهرت 7 أرقام لان لدينا 7 أيام

</div>

<div dir="rtl">

**توقعي (قبل التشغيل):**

- اتوقع ان يكون افضل فرع هو الرياض لان مجموع مبيعاته يبدو أعلى من بقية الفروع
- اتوقع ان يكون افضل يوم هو الجمعة لان إجمالي المبيعات فيه مرتفع جدًا عند جمع مبيعات جميع الفروع

</div>

```python
best_branch = branches[branch_totals.argmax()]
best_day = days[day_totals.argmax()]
branch_percentages = branch_totals / sales.sum() * 100

print("Best branch:", best_branch)
print("Best day:", best_day)
print("Branch percentages:", np.round(branch_percentages, 1))
```

<div dir="rtl">

**توقعي (قبل التشغيل):** اتوقع ان اعلى مبيعات ستكون في فرع الرياض يوم الجمعة، لان اعلى رقم ظاهر في الجدول هو 2410

</div>

```python
flat = sales.argmax()

row = flat // 7
col = flat % 7

print("Branch:", branches[row])
print("Day:", days[col])
print("Sales:", sales[row, col])
```

<div dir="rtl">

**توقعي (قبل التشغيل):** اتوقع ان نسبة الخسارة عند إغلاق فرع الدمام يوم الأحد حوالي 4% من مبيعات الأسبوع

</div>

```python
print(sales[2, 0] / sales.sum() * 100)
```

<div dir="rtl">

كان توقعي غير صحيح. توقعت تكون نسبة الخسارة حوالي 4%، لكن النتيجة الفعلية كانت حوالي 2.4% فقط. لذلك كان تقديري أعلى من الواقع، وظهر أن الخسارة أقل مما توقعت

</div>

<div dir="rtl">

**القرار:**

كما أن الخسارة المتوقعة من اغلاقه تمثل حوالي 2.4% فقط من إجمالي مبيعات الأسبوع، وهي نسبة منخفضة مقارنةً بإغلاق فرع آخر

**توقعي (قبل التشغيل):** اتوقع وجود 5 قراءات خاطئة

```python
depth = np.array([120, 340, 610, -1, 890, 1150, 1420, 99999, 1730, 2010,
                  2280, -1, 2560, 2840, 99999, 3120, 3390, 3650, -1, 3900])

mask = (depth == -1) | (depth == 99999)

print("Invalid readings:", mask)
print("Count:", mask.sum())
print("Percentage:", mask.sum() / len(depth) * 100)
```

<div dir="rtl">

**توقعي (قبل التشغيل):**

اتوقع يكون المتوسط مع القراءات الخاطئة اكبر من المتوسط بدونها. السبب هو وجود قيمتين كبيرتين (99999)، وهم اعلى بكثير من بقية الاعماق، لذلك سترفعان المتوسط بشكل كبير، بينما تأثير القيم -1 سيكون أقل

</div>

```python
good = depth[~mask]
```

```python
print("Mean with errors:", depth.mean())

print("Mean without errors:", good.mean())

print("Error multiplier:", depth.mean() / good.mean())
```

**بعد التشغيل:** توقعي كان صحيحًا. المتوسط مع القراءات الخاطئة (11500.25) أكبر بكثير من المتوسط بدونها (2000.67)، اي تضاعف الخطأ حوالي 5.75 مرة
