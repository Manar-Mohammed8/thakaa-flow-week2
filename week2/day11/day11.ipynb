### توقع خطوة 8: Outliers (IQR)

اتوقع وجود قيم متطرفه في الاعمده الثلاثة ويكون عددها الاكبر في `ApplicantIncome` حوالي 50 قيمة، وبعده `CoapplicantIncome` و`LoanAmount` حوالي 20 قيمة لكل واحد

</div>

```python
for col in ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]:
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    print(col, "→", count)
```

```python
for col in ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]:
    print(col)
    print(df_clean[col].sort_values(ascending=False).head(5))
```
